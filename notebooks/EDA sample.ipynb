```python
import pandas as pd
import numpy as np
import dask.dataframe as dd


# Set display options for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
csv_path = "../data/cleaned_sample.csv"

df = dd.read_csv(csv_path, blocksize="256MB", sample=10_000_000, encoding="utf-8")  # Increase sample size
print(df.head())  # Check if it loads correctly
```

```python
print(f"Information: {df.info()}")
print(df.shape)
print(df.size)
```

```python
#  Check Columns and Data Types
print(df.columns)  # Check column names
print(df.dtypes)   # Check data types
```

```python
# Compute missing values column by column
missing_counts = df.isnull().sum().compute(scheduler="processes")  # Use multiple CPU cores
print("Missing Values:\n", missing_counts)
```
